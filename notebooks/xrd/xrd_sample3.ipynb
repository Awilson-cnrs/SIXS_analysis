```python
%matplotlib inline
plt.close()

conditions = ['N2_RT', 'H2_RT', 'H2_200C', 'H2_400C', 'H2_200C_back', 'Mix_200C']

plt.figure(figsize=(10,5))

for condition in conditions:
    nxs = XRDdict[condition]['refScans'][0]
    h5Path = [file_path for file_path in homePath.rglob(f"*{nxs}-{nxs}*")][0]
    h5file = bu3.hdf5load(homePath/'binoculars/', h5Path)
    h5file.prjaxe('q')
    label = f'#{nxs} - {XRDdict[condition]['label']}'
    plt.plot(h5file.tth, h5file.img, label=label)

plt.legend()
```
